**Now I am generating the dataset**

```python
def generate_data(n=15): # a sample dataset of at least 15 viewing records
    data = []
    for _ in range(n):
        genre = random.choice(genres)
        watchtime = random.randint(60,80) #minutes
        data.append((genre, watchtime))
    return data
```
