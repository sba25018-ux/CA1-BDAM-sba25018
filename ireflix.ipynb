**Now I am defining genres**

```python
genres = ["Action", "Comedy", "Drama", "Horror"]
```

**I am now generating the dataset**

```python
def generate_data(n=15):
  data = []
  for _ in range(n):
    genre = random.choice(genres)
    watchtime = random.randint(60,80) #minutes
    data.append((genre, watchtime))
  return data
```

**Creating the dataset**

```python
data = generate_data(15)
```
