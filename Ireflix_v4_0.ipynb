**I am now asking Python to give me access to tools that can generate random numbers and random selections**

```python
import random
```

**I am now going to ask Python to generate a dataset of 15 records and 4 genres using Faker**

```python
fake = Faker
```

**Now I am defining genres**

```python
genres = ["Action", "Comedy","Drama","Horror"] # a minimum of 4 different genres
```

**Now I am generating the dataset**

```python
def generate_data(n=15): # a sample dataset of at least 15 viewing records
    data = []
    for _ in range(n):
        genre = random.choice(genres)
        watchtime = random.randint(60,80) #minutes
        data.append((genre, watchtime))
    return data
```

**Creating the dataset**

```python
data = generate_data(15) # a sample dataset of at least 15 viewing records
```

**Mapper Function**

```python
def mapper(data):
    mapped = []
    for record in data:
        genre, watchtime = record
        mapped.append((genre, watchtime))
    return mapped
```

**Reducer Function**



```python
def reducer(mapped_data):
    reduced = {}
    for genre, count in mapped_data:
        if genre in reduced:
            reduced[genre] += count
        else:
            reduced[genre] = count
    return reduced
```

**Main Function**

```python
def main():
    mapped = mapper(data)
    reduced = reducer(mapped)
    print("View Count per Genre:")
    for genre, count in reduced.items():
        print(f"{genre}: {count}")
```

**Run the solution**

```python
if __name__ == "__main__":
    main()
```
